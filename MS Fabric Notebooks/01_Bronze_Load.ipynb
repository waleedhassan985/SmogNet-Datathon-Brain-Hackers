```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
# ============================================
# Read Training files
# ============================================

training_df = spark.read \
.option("header","true") \
.option("inferSchema","true") \
.csv("Files/Bronze/Training/*.csv")

# Add metadata columns
training_df = training_df.withColumn(
    "dataset_type",
    lit("Training")
)

training_df = training_df.withColumn(
    "source_file",
    input_file_name()
)

print("Training rows loaded:")

print(training_df.count())

display(training_df.limit(5))
```

```python
# ============================================
# Read Testing files
# ============================================

testing_df = spark.read \
.option("header","true") \
.option("inferSchema","true") \
.csv("Files/Bronze/Testing/*.csv")

testing_df = testing_df.withColumn(
    "dataset_type",
    lit("Testing")
)

testing_df = testing_df.withColumn(
    "source_file",
    input_file_name()
)

print("Testing rows loaded:")

print(testing_df.count())

display(testing_df.limit(5))
```

```python
print("Training Schema")

training_df.printSchema()

print("Testing Schema")

testing_df.printSchema()
```

```python
print("Training Missing Values")

display(training_df.select([sum(col(c).isNull().cast("int")).alias(c)
for c in training_df.columns]))
```

```python
# ============================================
# Save Training
# ============================================

training_df.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable("Bronze_training")


# ============================================
# Save Testing
# ============================================

testing_df.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable("Bronze_testing")

print("Bronze tables created successfully")
```

```python
bronze_train=spark.sql("""

SELECT *
FROM Bronze_training
LIMIT 10

""")

display(bronze_train)
```

```python
bronze_test=spark.sql("""

SELECT *
FROM Bronze_testing
LIMIT 10

""")

display(bronze_test)
```
